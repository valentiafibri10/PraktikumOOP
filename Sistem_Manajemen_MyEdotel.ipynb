```python
from abc import ABC, abstractmethod

#ABSTRACTION
class KamarHotel(ABC):
    # setiap child class WAJIB mengisi ini (dipakai untuk pesan setup stok,
    # supaya tidak tercampur dengan nama spesifik kamar di 'nama_kamar')
    TIPE = "Kamar"

    def __init__(self, nama_kamar, harga_dasar, stok):
        # ENCAPSULATION -> atribut sensitif diberi tanda __ (private)
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar
        self.__stok = stok

    # ----- GETTER -----
    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    # ----- SETTER / method tambah_stok dengan validasi -----
    def tambah_stok(self, jumlah):
        stok_baru = self.__stok + jumlah
        if stok_baru < 0:
            print(f"Gagal update stok {self.TIPE}! Stok tidak boleh negatif ({stok_baru}).")
            return False
        else:
            self.__stok = stok_baru
            print(f"Berhasil menambahkan stok {self.TIPE}: {self.__stok} unit.")
            return True
    # ----- ABSTRACT METHOD (wajib di-override oleh child class) -----
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass
```

```python
# INHERITANCE + POLYMORPHISM
class KamarDeluxe(KamarHotel):
    TIPE = "Kamar Deluxe"
    PAJAK = 0.10  # 10%

    def __init__(self, nama_kamar, harga_dasar, stok, fasilitas):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * self.PAJAK
        print(f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"   Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak(10%): Rp {pajak:,.0f}".replace(",", "."))

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() * (1 + self.PAJAK)
        return harga_per_malam * jumlah_malam
```

```python
class KamarStandard(KamarHotel):
    TIPE = "Kamar Standard"
    PAJAK = 0.05  # 5%

    def __init__(self, nama_kamar, harga_dasar, stok, kapasitas):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * self.PAJAK
        print(f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"   Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak(5%): Rp {pajak:,.0f}".replace(",", "."))

    def hitung_harga_total(self, jumlah_malam):
        harga_per_malam = self.get_harga_dasar() * (1 + self.PAJAK)
        return harga_per_malam * jumlah_malam
```

```python
def proses_transaksi(daftar_pesanan):
    """
    daftar_pesanan: list of tuple (objek_kamar, jumlah_malam)
    """
    total_tagihan = 0
    print("--- STRUK PEMESANAN ---")
    for i, (kamar, jumlah_malam) in enumerate(daftar_pesanan, start=1):
        subtotal = kamar.hitung_harga_total(jumlah_malam)
        print(f"{i}. ", end="")
        kamar.tampilkan_detail()
        print(f"   Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(",", "."))
        total_tagihan += subtotal

    print("-" * 40)
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}".replace(",", "."))
    print("-" * 40)
    return total_tagihan
```

```python
if __name__ == "__main__":
    print("--- SETUP DATA KAMAR ---")

    # a) Admin membuat data kamar
    deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1_500_000, 0, "Private Pool")
    standard = KamarStandard("Kamar Standard Superior", 500_000, 0, "2 Orang")

    # b) Admin menambah stok (termasuk mencoba angka negatif -> harus ditolak)
    deluxe.tambah_stok(10)          # berhasil
    standard.tambah_stok(-5)        # ditolak, karena stok jadi negatif
    standard.tambah_stok(20)        # berhasil

    # c) Tamu memesan: 2 malam Deluxe, 1 malam Standard
    daftar_pesanan = [
        (deluxe, 2),
        (standard, 1),
    ]

    # d) Tampilkan struk & total tagihan
    proses_transaksi(daftar_pesanan)

    # bukti encapsulation: uncomment baris di bawah -> akan Error
    # print(deluxe.__harga_dasar)

    # bukti abstraction: uncomment baris di bawah -> akan Error (Can't instantiate abstract class)
    # kamar_gagal = KamarHotel("Test", 100000, 5)
```
